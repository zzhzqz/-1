```python
def f(x):        #计算反弹高度
    if x==1:
       return 50
    return 0.5*f(x-1)

def h(x):       #累计路程
    if x==1:
       return 150
    return h(x-1)+f(x-1)+f(x)

def t(x):     #总共用时
    if x==1:
        return (2*100/9.8)**0.5+(2*50/9.8)**0.5
    return t(x-1) + (2*f(x-1)/9.8)**0.5 + (2*f(x)/9.8)**0.5

print(f'第10次掉下并反弹到最高点时，反弹了{f(10)} 此时球一共经过{h(10)}米，运动了{t(10)}秒') 
n = int(input("请输入n的值："))
print(f'第{n}次掉下并反弹到最高点时，反弹了{f(n)} 此时球一共经过{h(n)}米，运动了{t(n)}秒') 
```
